```python
import pandas as pd
```

```python
df = pd.read_csv('article_topic_distribution_final_8.csv')
df.head()
```

```python
df = df.head(1000)
df.head()
```

compound_score	sentiment_label
>= 0.05	Positive 

<= -0.05	Negative 

(-0.05, 0.05)	Neutral

```python
sentiment_df = pd.read_csv('nbc_articles_cleaned_sentiment_flag_sample.csv')
sentiment_df = sentiment_df[['sentiment_label', 'compound_score']]
sentiment_df.head()
```

```python
concat_df = pd.concat([df, sentiment_df], axis=1)
concat_df.head()
```

```python
concat_df.to_csv('article_topic_distribution_with_sentiment_sample.csv', index=False)
```

```python

# 2️⃣ Convert datetime column to just date (remove time info)
concat_df['date'] = pd.to_datetime(concat_df['date']).dt.date

# 3️⃣ Identify topic columns dynamically
topic_cols = [col for col in concat_df.columns if col.startswith("topic_")]

# 4️⃣ Aggregate topic probabilities (mean per day)
daily_topics = (
    concat_df.groupby('date')[topic_cols]
      .mean()
      .reset_index()
      .sort_values('date')
)

# 5️⃣ Aggregate sentiment (min, max, mean)
daily_sentiment = (
    concat_df.groupby('date')['compound_score']
      .agg(min_sentiment='min',
           max_sentiment='max',
           avg_sentiment='mean')
      .reset_index()
)

# 6️⃣ Merge both
daily_combined_df = pd.merge(daily_topics, daily_sentiment, on='date', how='left')

# 7️⃣ Optional: round for readability
for col in topic_cols + ['min_sentiment', 'max_sentiment', 'avg_sentiment']:
    daily_combined_df[col] = daily_combined_df[col].round(6)

# 8️⃣ Preview result
print(daily_combined_df.head())

# 9️⃣ Save to CSV
daily_combined_df.to_csv("nbc_articles_cleaned_sentiment_topic_flag.csv", index=False)
```
